```python
def generate_wavelengths(
    start=400,
    end=2500,
    num_values=None,
    resolution=None,
    remove=None,
    keep=None
):
    """
    Generate a list of values with optional filtering.

    Parameters:
    - start (float): Start of the range.
    - end (float): End of the range.
    - num_values (int, optional): Number of values. If not provided, use resolution.
    - resolution (float, optional): Resolution. If not provided, use num_values.
    - remove (list, optional): List of 1-based indices to remove.
    - keep (list, optional): List of 1-based indices to keep.

    Returns:
    - list: Filtered list of values.
    """
    if num_values is not None:
        values = [start + i * (end - start) / (num_values - 1) for i in range(num_values)]
    elif resolution is not None:
        values = [start + i * resolution for i in range(int((end - start) / resolution))]
    else:
        raise ValueError("Either num_values or resolution must be provided.")

    if remove is not None:
        # Convert 1-based indices to 0-based
        remove = [idx - 1 for idx in remove]
        filtered_values = [val for idx, val in enumerate(values) if idx not in remove]

    elif keep is not None:
        filtered_values = [val for idx, val in enumerate(values) if idx in keep]

    else:
        filtered_values = values

    return filtered_values
```
